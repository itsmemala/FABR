```python
import json
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

### IMDB Movie

```python
data = pd.read_csv("dat\sent_mix\IMDB-Dataset.csv")
print(data.sentiment.unique())
data = data[:5000]
data = data.rename(columns={'review': 'sentence', 'sentiment': 'polarity'})
train_set, test_set = train_test_split(data,random_state=42,test_size=0.2,stratify=data['polarity'])
train_set, dev_set = train_test_split(train_set,random_state=42,test_size=0.1,stratify=train_set['polarity'])
print(train_set.shape, dev_set.shape, test_set.shape)
print((train_set['polarity']=='positive').sum(),(dev_set['polarity']=='positive').sum(),(test_set['polarity']=='positive').sum())
```

```python
with open("dat\\sent_mix\\Movie\\train.json", "w") as f:
    f.write(train_set.to_json(orient='index'))
with open("dat\\sent_mix\\Movie\\dev.json", "w") as f:
    f.write(dev_set.to_json(orient='index'))
with open("dat\\sent_mix\\Movie\\test.json", "w") as f:
    f.write(test_set.to_json(orient='index'))
```

### Twitter

```python
data = pd.read_csv("dat\\sent_mix\\twitter.csv", header=None)
data = data.rename(columns={0: 'polarity', 5: 'sentence'})
data = data[['polarity','sentence']]
data = data.groupby('polarity', group_keys=False).apply(lambda x: x.sample(2500))
print(data.polarity.unique())
data['polarity'] = np.where(data['polarity'] == 0, 'negative', 'positive')
print(data.polarity.unique())
train_set, test_set = train_test_split(data,random_state=42,test_size=0.2,stratify=data['polarity'])
train_set, dev_set = train_test_split(train_set,random_state=42,test_size=0.1,stratify=train_set['polarity'])
print(train_set.shape, dev_set.shape, test_set.shape)
print((train_set['polarity']=='positive').sum(),(dev_set['polarity']=='positive').sum(),(test_set['polarity']=='positive').sum())
```

```python
with open("dat\\sent_mix\\Twitter\\train.json", "w") as f:
    f.write(train_set.to_json(orient='index'))
with open("dat\\sent_mix\\Twitter\\dev.json", "w") as f:
    f.write(dev_set.to_json(orient='index'))
with open("dat\\sent_mix\\Twitter\\test.json", "w") as f:
    f.write(test_set.to_json(orient='index'))
```

### Finance

```python
data = pd.read_csv("dat\\sent_mix\\finance.csv", header=None)
data = data.rename(columns={0: 'polarity', 1: 'sentence'})
print(data.polarity.unique())
data = data[(data['polarity']=='positive') | (data['polarity']=='negative')]
print(data.polarity.unique())
train_set, test_set = train_test_split(data,random_state=42,test_size=0.2,stratify=data['polarity'])
train_set, dev_set = train_test_split(train_set,random_state=42,test_size=0.1,stratify=train_set['polarity'])
print(train_set.shape, dev_set.shape, test_set.shape)
print((train_set['polarity']=='positive').sum(),(dev_set['polarity']=='positive').sum(),(test_set['polarity']=='positive').sum())
```

```python
with open("dat\\sent_mix\\Finance\\train.json", "w") as f:
    f.write(train_set.to_json(orient='index'))
with open("dat\\sent_mix\\Finance\\dev.json", "w") as f:
    f.write(dev_set.to_json(orient='index'))
with open("dat\\sent_mix\\Finance\\test.json", "w") as f:
    f.write(test_set.to_json(orient='index'))
```
